2. Оскільки ми з'ясували, що жінки у місті Naypyitaw є нашим найбільш високочековим та найціннішим сегментом, тепер досліджуємо, що саме вони купують? Проаналізувавши які товари є популярними серед наших топових клієнтів, зможемо оптимізувати процес їхньої оплати.

```python
naypyitaw_women = df[(df['City'] == 'Naypyitaw') & (df['Gender'] == 'Female')]
naypyitaw_women.head(2)
```

```python
products_by_customer_type = naypyitaw_women.groupby(['Product line', 'Customer type'])['Invoice ID'].count().unstack().fillna(0)
products_by_customer_type['Total'] = products_by_customer_type.sum(axis=1)
products_by_customer_type.loc['Total'] = products_by_customer_type.sum(axis=0)
products_by_customer_type
```

Food and beverages (38 покупок) та Fashion accessories (35 покупок) є абсолютними фаворитами серед жінок у Naypyitaw.

Учасники програми лояльності купують товари категорії Home and lifestyle в тричі частіше (19/6). Food and beverages також показує суттєву перевагу серед лояльних клієнтів (23/15). Electronic accessories-лише 9 покупок роблять Member,коли звичайні покупці Normal купують їх удвічі частіше (19 покупок).

Відштовхуючтсь від результатів анадізу, можемо спробувати впровадити персоналізовані знижки та підвищений кешбек за карткою Member саме на категорії Food and beverages та Home and lifestyle. А для покупців без картки, тобто Normal, які купують Electronic accessories, можемо впровадити акцію на касі та запропонувати оформити картку лояльності прямо на касі та отримати бонусні бали на наступну покупку.
